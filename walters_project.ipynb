```python
import zipfile
import pandas as pd

# Unzip the file
zip_file_path = '/content/Medicare Telehealth Trends.zip'
with zipfile.ZipFile(zip_file_path, 'r') as zip_ref:
    zip_ref.extractall('/content/medicare_telehealth_data')

# The CSV file is located in a subfolder after extraction
csv_file_path = '/content/medicare_telehealth_data/Medicare Telehealth Trends/2025-Q4/TMEDTREND_PUBLIC_260610.csv'

# Load the data into a pandas DataFrame
df = pd.read_csv(csv_file_path)
```

```python
# Display the first 5 rows of the DataFrame
display(df.head())
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Group by 'Bene_Geo_Desc' (State) and calculate the mean of 'Pct_Telehealth'
telehealth_by_state = df.groupby('Bene_Geo_Desc')['Pct_Telehealth'].mean().reset_index()

# Sort by percentage for better visualization
telehealth_by_state_sorted = telehealth_by_state.sort_values(by='Pct_Telehealth', ascending=False)

# Create the bar plot
plt.figure(figsize=(12, 8))
sns.barplot(x='Pct_Telehealth', y='Bene_Geo_Desc', data=telehealth_by_state_sorted, palette='viridis')
plt.xlabel('Average Percentage of Telehealth Beneficiaries')
plt.ylabel('State / Geographical Description')
plt.title('Average Percentage of Telehealth Beneficiaries by State')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns

# Group by 'Bene_Geo_Desc' (State) and calculate the mean of 'Total_Bene_Telehealth'
telehealth_total_by_state = df.groupby('Bene_Geo_Desc')['Total_Bene_Telehealth'].mean().reset_index()

# Sort by total beneficiaries for better visualization
telehealth_total_by_state_sorted = telehealth_total_by_state.sort_values(by='Total_Bene_Telehealth', ascending=False)

# Create the bar plot
plt.figure(figsize=(12, 8))
sns.barplot(x='Total_Bene_Telehealth', y='Bene_Geo_Desc', data=telehealth_total_by_state_sorted, palette='viridis')
plt.xlabel('Average Total Telehealth Beneficiaries')
plt.ylabel('State / Geographical Description')
plt.title('Average Total Telehealth Beneficiaries by State')
plt.grid(axis='x', linestyle='--', alpha=0.7)
plt.tight_layout()
plt.show()
```

```python
# Reload the df_macs DataFrame to get the original data with string values
df_macs = pd.read_csv('/content/2025-all-99213-99214-all_macs.csv')

# Columns to convert to numeric
columns_to_convert = [
    'Non-Facility Price',
    'Facility Price',
    'Non-Facility Limiting Charge',
    'Facility Limiting Charge'
]

# Clean and convert specified columns to numeric
for col in columns_to_convert:
    # Remove '$' and any potential commas, then convert to numeric
    df_macs[col] = df_macs[col].astype(str).str.replace('$', '', regex=False).str.replace(',', '', regex=False)
    df_macs[col] = pd.to_numeric(df_macs[col], errors='coerce')

print("Data types after cleaning and conversion:")
df_macs[columns_to_convert].info()

print("\nUpdated DataFrame Descriptive Statistics (for converted columns):")
display(df_macs[columns_to_convert].describe())
```
